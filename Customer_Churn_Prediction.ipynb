```python
unique_colum=['Geography','Gender','NumOfProducts', 'HasCrCard', 'IsActiveMember']
```

```python
for co in unique_colum:
 print(df[co].unique())
```

```python
df = pd.get_dummies(df, columns=['Geography','Gender'], drop_first=True)
```

```python
df.head()
```

```python
df=df.astype(float)
```

```python
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
cols = ['CreditScore','Age','Tenure','Balance','NumOfProducts','EstimatedSalary']

df[cols] = scaler.fit_transform(df[cols])
```

```python
df.head()
```

```python
X=df.drop('Exited',axis=1)
y=df['Exited']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.33, random_state=42)
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
model=keras.Sequential([keras.layers.InputLayer(shape=(X_train.shape[1],)),
                        keras.layers.Dense(20,activation='relu'),
                        keras.layers.Dense(16,activation='relu'),
                        keras.layers.Dense(18,activation='relu'),
                        keras.layers.Dense(1,activation='sigmoid')])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train,y_train,epochs=200)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
y_predict=model.predict(X_test)
```

```python
accuracy_score(y_test,y_predict.round())
```
